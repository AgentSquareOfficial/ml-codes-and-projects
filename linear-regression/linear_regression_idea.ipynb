A simple code to show the working of the linear model...

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
print(f"x_train = {x_train}")
print(f"y_train = {y_train}")
```

```python
i = 1
x_i = x_train[i]
y_i = y_train[i]
print(f"(x^({i}), y^({i})) = ({x_i}, {y_i})")
```

```python
m = len(x_train)
```

```python
plt.scatter(x_train, y_train, marker="x", color="r")
plt.ylabel("size in feet^2")
plt.xlabel("price in $1000's")
plt.title("House sizes and prices")
plt.show()
```

after testing MANUALLY, for w = 200 and b = 100, the error in our small model is zero..

```python
w = 200
b = 100
```

```python
def compute_model_output(x, w, b):
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w*x[i]+b

    return f_wb
```

```python
temp_f_wb = compute_model_output(x_train, w, b)
plt.plot(x_train, temp_f_wb, c="b", label="Our Prediction")
plt.scatter(x_train, y_train, marker="x", c="r", label="Actual Values")
plt.xlabel("size in feet^2")
plt.ylabel("price in $1000's")
plt.title("House sizes and prices")
plt.legend()
plt.show()
```
