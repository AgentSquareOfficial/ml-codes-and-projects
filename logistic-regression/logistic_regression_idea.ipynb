```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    g = 1/(1+np.exp(-z))
    return g
```

```python
z_tmp = np.arange(-10, 11)

y = sigmoid(z_tmp)

fig, ax = plt.subplots(1, 1, figsize=(5,3))

ax.plot(z_tmp, y, c='blue')
ax.axvline(0, 0, 1, linestyle='--')
ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
```
