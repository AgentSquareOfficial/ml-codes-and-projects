Briefly explaining the decision boundary...

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])
```

```python
fig,ax = plt.subplots(1,1, figsize=(4,4))
pos = y==1
neg = y==0
x_0 = x[:,0]
x_1 = x[:,1]
ax.scatter(x_0[pos],x_1[pos],marker='x',c='red',lw=3,label='y=1')
ax.scatter(x_0[neg],x_1[neg],marker='o',lw=3,edgecolors='b',facecolors='none',label='y=0')
ax.axis([0, 4, 0, 3.5])
ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
ax.legend()

x0 = np.arange(0,6)
x1 = 3 - x0 # assuming w=1 and b=3 
ax.plot(x0,x1,c='b') # -3 + x0 + x1 = 0
ax.fill_between(x0,x1, alpha=0.2) 
plt.show()

```
