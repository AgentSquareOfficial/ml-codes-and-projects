## Logistic Regression, Logistic Loss idea

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([0, 1, 2, 3, 4, 5])
y_train = np.array([0, 0, 0, 1, 1, 1])

pos = y_train==1
neg = y_train==0

fig, ax = plt.subplots(1, 1, figsize=(6, 4))
ax.scatter(x_train[pos], y_train[pos], marker='x', s=80 ,c='red', lw=3, label='malignant')
ax.scatter(x_train[neg], y_train[neg], marker='o', s=80 ,facecolors='none', edgecolors='b', lw=2, label='benign')
ax.set_xlabel('Tumor Size')
ax.set_ylabel('y')
ax.set_title('Example of Logistic Regression on Categorical Data')
ax.legend()
plt.show()
```

```python
def sigmoid(z):
    g = 1/(1+np.exp(-z))
    return g
```

```python
def compute_cost(x, y, w, b):
    J_wb = 0
    m = x.shape[0]
    for i in range(m):
        J_wb += (sigmoid(w*x[i] + b) - y[i])**2
    J_wb = J_wb / (2 * m)
    return J_wb
```

```python
def plt_logistic_squared_error(x, y):
    w = np.linspace(-10, 12, 200)
    b = np.linspace(-20, 10, 200)

    W, B = np.meshgrid(w, b)
    J = np.zeros(W.shape)
    for i in range(W.shape[0]):
        for j in range(W.shape[1]):
            J[i, j] = compute_cost(x, y, W[i, j], B[i, j])

    fig = plt.figure(figsize=(10, 7))
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(W, B, J, cmap='rainbow', edgecolor='none')
    ax.set_xlabel('w')
    ax.set_ylabel('b')
    ax.set_zlabel('Cost')
    ax.set_title('Logistic Squared Error Cost vs (w, b)')
    plt.show()
```

```python
plt_logistic_squared_error(x_train, y_train)
```

### Showing how Loss function works...

```python
def plt_two_logistic_loss_curves():
    fig, ax = plt.subplots(1, 2, figsize=(9, 4))
    f_wb = np.linspace(0.1,0.9,100)
    loss = -np.log(f_wb)
    ax[0].plot(f_wb,loss,c='b')
    ax[0].set_xlabel('f_w,b(x)')
    ax[0].set_ylabel('loss')
    ax[0].set_title('y=1')

    loss1 = -np.log(1 - f_wb)
    ax[1].plot(f_wb,loss1,c='b')
    ax[1].set_xlabel('f_w,b(x)')
    ax[1].set_title('y=0')
    plt.plot()
```

```python
plt_two_logistic_loss_curves()
```

```python
def compute_cost_logistic(x, y, w, b):
    L_wb = 0
    m = x.shape[0]
    for i in range(m):
        L_wb += (-y[i]*np.log(sigmoid(w*x[i]+b)) - (1 - y[i])* np.log(1 - sigmoid(w*x[i]+b)))
    J_wb = L_wb / m
    return J_wb
```

```python
def plt_logistic_cost(x, y):
    w = np.linspace(-10, 12, 200)
    b = np.linspace(-20, 10, 200)

    W, B = np.meshgrid(w, b)
    J = np.zeros(W.shape)
    for i in range(W.shape[0]):
        for j in range(W.shape[1]):
            J[i, j] = compute_cost_logistic(x, y, W[i, j], B[i, j])

    fig = plt.figure(figsize=(12, 5))
    ax1 = fig.add_subplot(121, projection='3d')
    ax2 = fig.add_subplot(122, projection='3d')
    ax1.plot_surface(W, B, J, cmap='rainbow', edgecolor='none')
    ax1.set_xlabel('w')
    ax1.set_ylabel('b')
    ax1.set_zlabel('Cost')
    ax1.set_title('Logistic Cost vs (w, b)')

    ax2.plot_surface(W, B, np.log(J), cmap='rainbow', edgecolor='none')
    ax2.set_xlabel('w')
    ax2.set_ylabel('b')
    ax2.set_zlabel('Cost')
    ax2.set_title('log(Logistic Cost) vs (w, b)')
    plt.show()
```

```python
plt_logistic_cost(x_train, y_train)
```
