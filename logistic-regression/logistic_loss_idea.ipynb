## Logistic Regression, Logistic Loss idea

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([0, 1, 2, 3, 4, 5])
y_train = np.array([0, 0, 0, 1, 1, 1])

pos = y_train==1
neg = y_train==0

fig, ax = plt.subplots(1, 1, figsize=(6, 4))
ax.scatter(x_train[pos], y_train[pos], marker='x', s=80 ,c='red', lw=3, label='malignant')
ax.scatter(x_train[neg], y_train[neg], marker='o', s=80 ,facecolors='none', edgecolors='b', lw=2, label='benign')
ax.set_xlabel('Tumor Size')
ax.set_ylabel('y')
ax.set_title('Example of Logistic Regression on Categorical Data')
ax.legend()
plt.show()
```

```python
def sigmoid(z):
    g = 1/(1+np.exp(-z))
    return g
```

```python
def compute_cost(x, y, w, b):
    J_wb = 0
    m = x.shape[0]
    for i in range(m):
        J_wb += (sigmoid(w*x[i] + b) - y[i])**2
    J_wb = J_wb / (2 * m)
    return J_wb
```

```python
def plt_logistic_squared_error(x, y):
    w = np.linspace(-10, 12, 200)
    b = np.linspace(-20, 10, 200)

    W, B = np.meshgrid(w, b)
    J = np.zeros(W.shape)
    for i in range(W.shape[0]):
        for j in range(W.shape[1]):
            J[i, j] = compute_cost(x, y, W[i, j], B[i, j])

    fig = plt.figure(figsize=(10, 7))
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(W, B, J, cmap='rainbow', edgecolor='none')
    ax.set_xlabel('w')
    ax.set_ylabel('b')
    ax.set_zlabel('Cost')
    ax.set_title('Logistic Squared Error Cost vs (w, b)')
    plt.show()
```

```python
plt_logistic_squared_error(x_train, y_train)
```
